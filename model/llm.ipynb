### LLM에서 받은 3가지 지역 / 유저 입력 쿼리에서 뽑아낸 나이, 성별, 직무로 metadata 필터 만들기

- 성별, 직무에 관한 정책은 좀 애매한거 같기도 ?
- 나이, 소득으로 가장 많이 갈리지 않나

```python
# 유저 쿼리 파싱
def parse_user_query(query):
    parsed = {}
    parsed["age"] = int(re.search(r"나이:\s*(\d+)", query).group(1))
    parsed["gender"] = re.search(r"성별:\s*(\w+)", query).group(1)
    parsed["job_interest"] = re.search(r"관심 직무:\s*([\w\s]+)", query).group(1).strip()
    return parsed
```

```python
# 정책 필터링 함수
def get_matched_policies(user_info, region_list, policy_df):
    matched = []

    for region in region_list:
        # 해당 지역에 대한 정보라 정책DB에 지역이름이 들어가야 함.
        ## 근데 충북지역으로 일단 진행하는거니 괜찮을 것 같기도
        region_policies = policy_df[policy_df["region"] == region]

        # 나이 필터
        region_policies = region_policies[
            (region_policies["age_min"] <= user_info["age"]) &
            (region_policies["age_max"] >= user_info["age"])
        ]

        # 성별 필터
        ## 성별에 따른 정책이 많이 나뉜다고 명인이가 얘기하긴 했는데 저 DB에서 그걸로 나누긴 힘들 것 같기도
        region_policies = region_policies[
            (region_policies["gender"] == user_info["gender"]) |
            (region_policies["gender"] == "전체")
        ]

        # 직무 키워드 필터
        ## 직무도 사실 모르겠음 어떤 식으로 필터링 걸어서 정책 뿌려줘야 할지
        region_policies = region_policies[
            region_policies["job_keywords"].apply(lambda keywords: any(
                kw in user_info["job_interest"] for kw in keywords))
        ]

        matched.append(region_policies[["policy_name", "summary"]].to_dict(orient="records"))

    return matched
```

```python
# upload_to_pinecone.py
import json
from openai import OpenAI
from pinecone import Pinecone

# 환경 설정
client = OpenAI(api_key="OPENAI_API_KEY")
pc = Pinecone(api_key="PINECONE_API_KEY")
index = pc.Index("regions-index")

# JSON 로드
with open("regions.json", "r", encoding="utf-8") as f:
    regions = json.load(f)

# Pinecone에 업로드
for region in regions:
    text = region["summary_box"]
    emb = client.embeddings.create(
        input=text,
        model="text-embedding-3-small"
    ).data[0].embedding

    index.upsert([
        (
            str(region["id"]),
            emb,
            {
                "name": region["name"],
                "house_price": region["house_price"],
                "education": region["education"],
                "medical": region["medical"],
                "summary_box": region["summary_box"]
            }
        )
    ])

print("Pinecone 업로드 완료")
```

```python
# recommend.py
from openai import OpenAI
from pinecone import Pinecone

client = OpenAI(api_key="OPENAI_API_KEY")
pc = Pinecone(api_key="PINECONE_API_KEY")
index = pc.Index("regions-index")

def recommend_regions(user_input):
    """
    user_input 예시:
    {
        "house_price": "하",
        "education": "상",
        "medical": "중"
    }
    """
    # 사용자 입력을 텍스트로 변환
    query_text = f"""
    사용자가 원하는 조건:
    - 집값: {user_input['house_price']}
    - 교육: {user_input['education']}
    - 의료: {user_input['medical']}
    조건에 맞는 한국 지역을 추천하세요.
    """

    # 쿼리 임베딩
    emb = client.embeddings.create(
        input=query_text,
        model="text-embedding-3-small"
    ).data[0].embedding

    # Pinecone 검색 (Top 3)
    results = index.query(vector=emb, top_k=3, include_metadata=True)

    # 카드로 뿌리기 좋은 형태로 가공
    recommendations = []
    for match in results["matches"]:
        meta = match["metadata"]
        recommendations.append({
            "name": meta["name"],
            "house_price": meta["house_price"],
            "education": meta["education"],
            "medical": meta["medical"],
            "reason": meta["summary_box"]
        })

    return {"recommendations": recommendations}

# 테스트
if __name__ == "__main__":
    user_input = {
        "house_price": "하",
        "education": "상",
        "medical": "중"
    }
    print(recommend_regions(user_input))
```
